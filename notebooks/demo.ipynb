# JobMatch demo (Phase 1)

This notebook **imports** the real modules from `src/jobmatch`; no agent logic is duplicated here. Add your Gemini key to `.env` before running the run cells.

```python
from jobmatch.graph import build_graph, resume_run, start_run
from jobmatch.schemas import ApprovalDecision

# Uses .env for the API key and data/checkpoints.sqlite for durable state.
graph = build_graph()
thread_id = "demo-1"
```

```python
# The assistant calls the tools, drafts an assessment, then pauses.
paused = start_run(
    graph,
    thread_id=thread_id,
    message="Which of these jobs fits my CV best?",
    cv_path="data/sample_cv.pdf",
    job_links=["https://techny.dev/jobs/forward-deployed-engineer-ai-arabicai-b7a8eb"],
)
paused["__interrupt__"][0].value
```

A human reviews the draft above. In the API this is `POST /runs/{thread_id}/resume`; here we resume directly with `Command(resume=...)` under the hood.

```python
result = resume_run(graph, thread_id, ApprovalDecision(approved=True))
print(result["final_answer"])
```
